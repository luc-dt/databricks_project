# Select Catalog

```python
%sql
USE CATALOG workspace;
```

# Create Lakehouse Schemas

```python
%sql
CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Bronze layer: raw ingested data';

CREATE SCHEMA IF NOT EXISTS silver
COMMENT 'Silver layer: cleaned and transformed data';

CREATE SCHEMA IF NOT EXISTS gold
COMMENT 'Gold layer: business-ready data';
```

#Create Volume

```python
%sql
CREATE VOLUME IF NOT EXISTS workspace.bronze.raw_sources
COMMENT 'Volume for raw source files (CSV)';
```

#Create Monitoring Schema

```python
%sql
CREATE SCHEMA IF NOT EXISTS monitoring
COMMENT 'monitoring layer: ingestion data';
```

#Create Audit Log Table

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS workspace.monitoring.pipeline_audit_log (
    notebook_name STRING,
    target_table STRING,
    run_timestamp TIMESTAMP,
    rows_inserted LONG,
    rows_updated LONG,
    rows_deleted LONG,
    qc_status STRING,
    qc_message STRING
)
COMMENT 'Audit log for pipeline runs';
""")
```
